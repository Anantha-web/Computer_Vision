## Part 4

```python
import os
import shutil
import random

dataset_path = '/content/drive/MyDrive/archive/'
images_path = os.path.join(dataset_path, 'images')
train_images_path = os.path.join(images_path, 'train')
train_labels_path = os.path.join(dataset_path, 'labels', 'train')
config_path = os.path.join(dataset_path, 'config.yaml')

train1_path = os.path.join(dataset_path, 'train1')
os.makedirs(train1_path, exist_ok=True)
train1_images_path = os.path.join(train1_path, 'images', 'train')
train1_labels_path = os.path.join(train1_path, 'labels', 'train')
os.makedirs(train1_images_path, exist_ok=True)
os.makedirs(train1_labels_path, exist_ok=True)

train2_path = os.path.join(dataset_path, 'train2')
if os.path.exists(train2_path):
    shutil.rmtree(train2_path)

shutil.copytree(dataset_path, train2_path)

image_files = os.listdir(train_images_path)

selected_images = random.sample(image_files, 100)

for image in selected_images:
    image_path = os.path.join(train_images_path, image)
    label_path = os.path.join(train_labels_path, os.path.splitext(image)[0] + '.txt')
    shutil.copy(image_path, train1_images_path)
    shutil.copy(label_path, train1_labels_path)

with open(config_path, 'r') as f:
    config_data = f.read()

config_data = config_data.replace('images/train', 'train1/images/train').replace('labels/train', 'train1/labels/train')

train1_config_path = os.path.join(train1_path, 'config.yaml')
with open(train1_config_path, 'w') as f:
    f.write(config_data)

print("Dataset creation completed!")
```

### (a) How does increasing dataset size impact performance?

Increasing the dataset size in training YOLOv8 models leads to better performance, as evidenced by higher mean Average Precision (mAP) values on both training and validation sets. This improvement indicates that the models become more adept at accurately detecting objects and assigning higher confidence scores to them.

By training on a larger dataset, the models are exposed to a wider variety of examples, including different object types, variations in appearance, backgrounds, lighting conditions, and other factors. This exposure helps the models generalize better to unseen data, capturing a broader range of complexities present in real-world images.

Even when starting from pretrained weights, increasing the dataset size still leads to performance improvements. This suggests that while pretrained models provide a good starting point, fine-tuning on a larger dataset further enhances their performance by adapting them to specific data characteristics and requirements.

### (b) Does the bigger model perform better? Why or why not?

The comparison between YOLOv8m and YOLOv8n models sheds light on the complex relationship between model size and performance. While YOLOv8m, being a larger model, achieves higher mean Average Precision (mAP) on the training set, it exhibits signs of overfitting by performing worse on the validation set compared to YOLOv8n. Overfitting occurs when a model learns to memorize the training data instead of generalizing well to unseen data. This suggests that while larger models have the capacity to capture more complex patterns in the data, they may also be more susceptible to overfitting, especially when the dataset size is limited.

Moreover, the performance of a model depends on a multitude of factors beyond just its size, including the quality of pretrained weights, dataset size, and task complexity. For instance, a smaller model with carefully initialized weights or trained on a larger and more diverse dataset might outperform a larger model in certain scenarios. Therefore, it's essential to conduct thorough empirical evaluations, experimenting with different model architectures, dataset sizes, and training strategies to identify the most suitable model for a specific application.

In conclusion, the comparison between YOLOv8m and YOLOv8n underscores the importance of striking a balance between model complexity and generalization ability. It highlights the need for careful consideration of various factors and meticulous experimentation to choose the optimal model that best addresses the requirements and challenges of a given task.

#### (i) Does the default run use augmentations? Which ones and what do they do?

The default training setup includes data augmentation using the Albumentations library, incorporating the following augmentations:

Blur: Introduces blur to images with a probability of 0.01 and a blur limit between 3 and 7.
MedianBlur: Applies median blur with a probability of 0.01 and a blur limit between 3 and 7.
ToGray: Converts images to grayscale with a probability of 0.01.
CLAHE: Utilizes Contrast Limited Adaptive Histogram Equalization (CLAHE) with a probability of 0.01, a clip limit ranging from 1 to 4.0, and a tile grid size of 8x8.
These augmentations are incorporated into the training process to enhance the model's robustness by introducing variations in the training images.

### (iii) Which augmentation is the most important? How did you determine this?

Certain augmentation techniques are commonly considered more important or impactful based on their ability to introduce variations that help improve model generalization.

- Horizontal Flipping: Helps the model learn invariant features and doubles the dataset size.
- Rotation: Introduces robustness to rotational changes in real-world scenarios.
- Scaling and Translation: Makes the model invariant to changes in scale and position.
- Color Jittering: Helps the model become invariant to variations in lighting conditions and color distributions.
- Random Cropping: Introduces variability in the spatial location of objects, aiding the model in recognizing objects in different contexts and scales.
